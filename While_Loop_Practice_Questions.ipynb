**While Loop – 10 Interview Questions**

**Basic While Loop**

1.	Print numbers from 1 to 10.
2.	Print all even numbers between 1 and 50.
3.	Print the sum of numbers from 1 to 100.
4.	Take user input repeatedly until they type "exit".
5.	Print the multiplication table of a given number.
6.	Count the number of digits in a number.
7.	Reverse a number entered by the user.
8.	Find the factorial of a given number.
9.	Print Fibonacci numbers less than 50.
10.	Find the largest number in a list using while.



```python
# Basic While Loop

## Print numbers from 1 to 10.

i = 1

while i < 11:
  print(i)

  i += 1
```

```python
## Print all even numbers between 1 and 50.

i = 2

while i <= 50:
  print(i)

  i += 2
```

```python
## Print the sum of numbers from 1 to 100.

i = 1
total = 0

while i <= 100:
  total = total + i
  i = i+1

print("Sum: ", total)
```

```python
## Take user input repeatedly until they type "exit".

while True:
  user_input = input("Enter something (type 'exit' to Stop): ")

  if user_input.lower() == "exit":
    break

  print("You entered: ", user_input)
```

```python
## Print the multiplication table of a given number.

number = int(input("Enter the number: "))

i = 1
while i <= 10:
  print(number, "*", i, "=", number*i)

  i += 1
```

```python
## Count the number of digits in a number.

number = int(input("Enter the number: "))

count = 0

while number > 0:
  number = number // 10
  count += 1

print("Number of digits: ", count)
```

```python
## Reverse a number entered by the user.

number = int(input("Enter the number: "))

reverse = 0

while number > 0:
  digit = number % 10
  reverse = reverse * 10 + digit
  number = number // 10

print("Reversed number: ", reverse)
```

```python
# Find the factorial of a given number.

number = int(input("Enter a number: "))

factorial = 1
i = 1

while i <= number:
    factorial = factorial * i
    i = i + 1

print("Factorial of", number, "is", factorial)
```

```python
# Print Fibonacci numbers less than 50.

# Fibonacci numbers are a special sequence of integers where each number is the sum of the two preceding ones, starting from 0 and 1

a = 0
b = 1

while a < 50:
    print(a)
    a, b = b, a + b
```

```python
# Find the largest number in a list using while.

numbers = [25, 10, 75, 40, 90, 55]

largest = numbers[0]
i = 1

while i < len(numbers):
    if numbers[i] > largest:
        largest = numbers[i]
    i += 1

print("Largest number is:", largest)
```

**While Loop with break and continue**

1.	Ask the user to enter numbers repeatedly; stop when they enter 0 (break).
2.	Print numbers 1 to 20 but skip 5 (continue).
3.	Keep asking for a password until correct; exit loop on correct input (break).
4.	Print numbers 1 to 30 but skip multiples of 3 (continue).
5.	Print numbers 1 to 15 but stop if number > 10 (break).
6.	Take continuous user input and ignore negative numbers (continue).
7.	Print squares of numbers 1 to 20 but stop when square > 100 (break).
8.	Print numbers divisible by 4 between 1 and 50 but skip 16 (continue).
9.	Keep reading numbers until sum exceeds 100 (break).
10.	Print numbers from 1 to 20 but skip multiples of 5 (continue).

```python
## While Loop with break and continue

# Ask the user to enter numbers repeatedly; stop when they enter 0 (break).

while True:
    number = int(input("Enter a number (0 to stop): "))

    if number == 0:
        print("You entered 0. Loop stopped.")
        break

    print("You entered:", number)
```

```python
# Print numbers 1 to 20 but skip 5 (continue).

i = 1

while i <= 20:
    if i == 5:
        i += 1
        continue

    print(i)
    i += 1
```

```python
# Keep asking for a password until correct; exit loop on correct input (break).

correct_password = "Python123"

while True:
  password = input("Enter the Password: ")

  if password == correct_password:
    print("Correct Password: Access Granted!")
    break
  else:
    print("Incorrect Password. Try again:(")

```

```python
# print numbers 1 to 30 but skip multiples of 3 (continue).

i = 1

while i <= 30:
    if i%3 == 0:
        i += 1
        continue

    print(i)
    i += 1
```

```python
# Print numbers 1 to 15 but stop if number > 10 (break).

i = 1

while i <= 15:
  if i > 10:
    break
  print(i)
  i += 1
```

```python
# Take continuous user input and ignore negative numbers (continue).

while True:
    number = int(input("Enter a number (0 to stop): "))

    if number == 0:
        print("Loop stopped.")
        break

    if number < 0:
        continue

    print("You entered:", number)
```

```python
# Print squares of numbers 1 to 20 but stop when square > 100 (break).

i = 1

while i <= 20:
    square = i ** 2

    if square > 100:
        break

    print(i, "square =", square)
    i += 1
```

```python
# Print numbers divisible by 4 between 1 and 50 but skip 16 (continue).

i = 1

while i <= 50:
  if i == 16:
    i += 1
    continue

  if i % 4 == 0:
    print(i)

  i += 1
```

```python
# Keep reading numbers until sum exceeds 100 (break).

total = 0

while True:
    number = int(input("Enter a number: "))
    total += number

    print("Current sum:", total)

    if total > 100:
        print("Sum exceeded 100. Loop stopped.")
        break
```

```python
# Print numbers from 1 to 20 but skip multiples of 5 (continue).

i = 1

while i <= 20:
    if i%5 == 0:
        i += 1
        continue

    print(i)
    i += 1
```
